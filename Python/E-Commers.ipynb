```python
df.drop(["review_sentiment","customer_rating"],axis=1,inplace=True)
```

```python
# /fill delivery_days in blank celles in avg_delivery value
df['delivery_days'] = df['delivery_days'].fillna(
    df['delivery_days'].mean().round(0)
).astype(int)
```

```python
 # fill estimated_delivery_days with the avg estimated_delivery_days in blank cells 
df['estimated_delivery_days'] =df['estimated_delivery_days'].fillna(
    df['estimated_delivery_days'].mean().round(0)
).astype(int)
```

```python
# filling blank rows in return_statues with no return
df['return_status'] = df['return_status'].fillna('No Return')
```

```python
# filling blank rows in return_reason with no reason
df['return_reason']=df['return_reason'].fillna('No Reason')
```

```python
# fill avg value in customer rating in blank cells
df['customer_rating']=df['customer_rating'].fillna(
    df['customer_rating'].mean().round(1)
).astype(int)
```

```python
df.drop('customer_review', axis=1, inplace=True)
```

```python
# fill blank cells in review_sentiment 
df['review_sentiment'] = df['review_sentiment'].fillna('No Review')
```

```python
# fill blank campaign_name in unknown 
df['campaign_name'] =df['campaign_name'].fillna('Un known')
```

```python
# fill blank cells in  coupon_code 
df['coupon_code']=df['coupon_code'].fillna('No Coupon Use')
```

```python
df.to_csv('E-Commeres-Clean-Data set.csv')
```

```python
import numpy as np

conditions = [
    (df['customer_age'] >= 18) & (df['customer_age'] <= 25),
    (df['customer_age'] >= 26) & (df['customer_age'] <= 35),
    (df['customer_age'] >= 36) & (df['customer_age'] <= 45),
    (df['customer_age'] >= 46) & (df['customer_age'] <= 55),
    (df['customer_age'] >= 56) & (df['customer_age'] <= 65),
    (df['customer_age'] >= 66) & (df['customer_age'] <= 74)
]

choices = [
    'Young Adult',
    'Adult',
    'Mid Adult',
    'Mature Adult',
    'Senior',
    'Elderly'
]

df['Age_Category'] = np.select(
    conditions,
    choices,
    default='Unknown'
)
```
